```python

def validate_sequence():
    sequence = input("Enter DNA/RNA sequence= ").upper()

    if sequence == "":
        print("Empty sequence. Please enter a valid DNA or RNA sequence.")

    elif all(n in "ATGC" for n in sequence):
        print("Valid DNA sequence")
        return sequence, "DNA"

    elif all(n in "AUGC" for n in sequence):
        print("Valid RNA sequence")
        return sequence, "RNA"

    else:
        print("Invalid sequence. Please enter a valid DNA or RNA sequence.")

validate_sequence()

```
