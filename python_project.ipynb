```python
stop_codons = ["TAA", "TAG", "TGA"]

sequence = input("Enter DNA sequence: ").upper().strip()

found = False

for i in range(0, len(sequence) - 2, 3):
    codon = sequence[i:i+3]

    if codon in stop_codons:
        print("Stop codon found:", codon)
        found = True
        break

if not found:
    print("No stop codon found.")
```

```python
sequence = input("Enter DNA sequence: ").upper().strip()

if all(base in "ATGC" for base in sequence) and sequence != "":
    codons = []

    for i in range(0, len(sequence) - 2, 3):
        codons.append(sequence[i:i+3])

    print("Total complete codons:", len(codons))

    for codon in sorted(set(codons)):
        print(codon, "appears", codons.count(codon), "time(s)")
else:
    print("Please enter a valid DNA sequence.")
```

```python

def validate_sequence(sequence):
    sequence = sequence.upper().strip()

    if sequence == "":
        print("Empty sequence. Please enter a valid DNA or RNA sequence.")

    elif all(n in "ATGC" for n in sequence):
        print("Valid DNA sequence")
        return sequence, "DNA"

    elif all(n in "AUGC" for n in sequence):
        print("Valid RNA sequence")
        return sequence, "RNA"

    else:
        print("Invalid sequence. Please enter a valid DNA or RNA sequence.")

validate_sequence(sequence)

```

```python
def translate_codons(seq_2):
    seq_2 = seq_2.upper().strip()

    if seq_2 == "":
        print("No sequence provided.")
        return

    elif all(n in "ATGC" for n in seq_2):
        print("Valid DNA sequence")
        sequence_type = "DNA"

    elif all(n in "AUGC" for n in seq_2):
        print("Valid RNA sequence")
        sequence_type = "RNA"

    else:
        print("Invalid sequence. Please enter a valid DNA or RNA sequence.")
        return

    if len(seq_2) % 3 != 0:
        print("Warning: Incomplete codon at the end of the sequence will be ignored.")

    for i in range(0, len(seq_2) - 2, 3):
        codon = seq_2[i:i+3]

        if sequence_type == "RNA":
              codon = codon.replace("U", "T")

        amino_acid = codon_table[codon]
        print(codon, ":", amino_acid)

        if amino_acid == "Stop":
            print("Translation terminated due to stop codon.")
            break
    
translate_codons(sequence)
```
